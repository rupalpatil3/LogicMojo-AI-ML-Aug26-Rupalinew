## Problem: 4

Create a 10 x 10 arrays of zeros and then "frame" it with a border of ones.

```python
import numpy as np

arr = np.zeros((10, 10), dtype=int)

# Top and bottom borders
arr[0, :] = 1
arr[-1, :] = 1

# Left and right borders
arr[:, 0] = 1
arr[:, -1] = 1

print(arr)

```

## Problem: 5

Create an 8 x 8 array with a checkerboard pattern of zeros and ones.

```python
import numpy as np

checkerboard = np.zeros((8, 8), dtype=int)

checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1

print(checkerboard)

```

## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
import numpy as np

def create_array(n):
    i = np.arange(n).reshape(n, 1)  # column indices
    j = np.arange(n)                # row indices
    return i + j

# Example
print(create_array(4))

```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
import numpy as np

# 1000 uniformly spaced points in [0, 1]
x = np.linspace(0, 1, 1000)

# Evaluate cos(x) and sin(x)
cos_x = np.cos(x)
sin_x = np.sin(x)

# Stack into a tall array
A = np.column_stack((cos_x, sin_x))

print(A)
print(A.shape)
```

# Problem: 9

Create a random $3×5$ array using the `np.random.rand(3 ,5)` function and compute: the sum of all the entries, the sum of the rows, and the sum of the columns.

```python
import numpy as np

# Create a random 3 × 5 array
A = np.random.rand(3, 5)

print("Array:")
print(A)

# Sum of all entries
total_sum = np.sum(A)

# Sum of each row
row_sums = np.sum(A, axis=1)

# Sum of each column
column_sums = np.sum(A, axis=0)

print("\nSum of all entries:")
print(total_sum)

print("\nSum of rows:")
print(row_sums)

print("\nSum of columns:")
print(column_sums)

```

## Problem: 10

Create a random $5×5$ array using the function `np.random.rand(5, 5)`. Then sort the rows according to the second column.

```python
import numpy as np

# Create a random 5 × 5 array
A = np.random.rand(5, 5)

print("Original array:")
print(A)

# Sort rows by the second column (index 1)
A_sorted = A[np.argsort(A[:, 1])]

print("\nSorted array:")
print(A_sorted)

```
